# `data/streamlit_predictions.pkl` 내용 조회 노트북

Streamlit 대시보드(`app.py`)에서 사용하는 예측 데이터 파일 `data/streamlit_predictions.pkl`의 전체 구조와 각 항목의 내용을 확인하기 위한 노트북입니다.

## 1. 파일 로드 및 전체 키(Key) 구조 요약

```python
import os
import pickle
import pandas as pd
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', 1000)

PKL_PATH = 'data/streamlit_predictions.pkl'
file_size_mb = os.path.getsize(PKL_PATH) / (1024 * 1024)

with open(PKL_PATH, 'rb') as f:
    data = pickle.load(f)

print(f'파일 경로: {PKL_PATH} ({file_size_mb:.2f} MB)')
print(f'최상위 자료형: {type(data).__name__} (총 {len(data)}개 키)')

summary_rows = []
for k, v in data.items():
    if isinstance(v, pd.DataFrame):
        size_info = f'{v.shape[0]:,} rows x {v.shape[1]} cols'
    elif isinstance(v, (list, set, tuple, dict)):
        size_info = f'{len(v):,} items'
    else:
        size_info = str(v)[:50]
    summary_rows.append({
        'Key': k,
        'Type': type(v).__name__,
        'Size / Shape': size_info,
    })

display(pd.DataFrame(summary_rows))
```

## 2. `variety_order` (품종 순서) & `offseason_summary_df` (출하기·비수기 요약)

```python
print('=== variety_order ===')
print(data['variety_order'])

print('\n=== offseason_summary_df ===')
display(data['offseason_summary_df'])
```

## 3. `top5_units` (품종별 주요 거래단위 16개 그룹)

```python
top5_units = data['top5_units']
print(f'Shape: {top5_units.shape}')
display(top5_units)
```

## 4. `all_preds_df` (일자별 실적 및 4개 모델 예측 결과)

```python
all_preds_df = data['all_preds_df']
print(f'Shape: {all_preds_df.shape}')
print(f"전체 기간: {all_preds_df['date'].min()} ~ {all_preds_df['date'].max()}")

print('\n[컬럼 목록]')
for i, col in enumerate(all_preds_df.columns, 1):
    print(f'  {i:2d}. {col} ({all_preds_df[col].dtype})')

print('\n[과거(is_future=False) vs 미래(is_future=True) 구간 요약]')
period_summary = (
    all_preds_df.groupby(['variety', 'is_future'], observed=False)
    .agg(
        행수=('date', 'size'),
        시작일=('date', 'min'),
        종료일=('date', 'max'),
        그룹수=('group_id', 'nunique'),
    )
    .reset_index()
)
display(period_summary)

print('\n[상위 10행 (head)]')
display(all_preds_df.head(10))

print('\n[하위 10행 (tail)]')
display(all_preds_df.tail(10))
```

## 5. `all_preds_df` 결측치 및 주요 수치 컬럼 기초 통계량

```python
null_counts = all_preds_df.isna().sum()
null_df = pd.DataFrame({
    '결측치_수': null_counts,
    '결측률(%)': (null_counts / len(all_preds_df) * 100).round(2),
})
print('[결측치가 존재하는 컬럼]')
display(null_df[null_df['결측치_수'] > 0])

print('\n[주요 수치 컬럼 기초 통계량]')
num_cols = [
    'actual_vol_ton', 'actual_prc_kg', 'hist_vol_21_25', 'hist_prc_21_25',
    'TFT_vol_p50', 'XGBoost_vol_p50', 'LightGBM_vol_p50', 'RandomForest_vol_p50',
    'TFT_prc_p50', 'XGBoost_prc_p50', 'LightGBM_prc_p50', 'RandomForest_prc_p50',
]
display(all_preds_df[num_cols].describe().round(2).T)
```

## 6. `wk_full_grid` (그룹별·주차별 예년 기준 프로파일)

```python
wk_full_grid = data['wk_full_grid']
print(f'Shape: {wk_full_grid.shape}')
print('Columns:', list(wk_full_grid.columns))
display(wk_full_grid.head(15))
display(wk_full_grid.describe().round(2).T)
```

## 7. `active_season_pairs` (출하기 품종-주차) & `korean_holidays` (공휴일 목록)

```python
from collections import defaultdict

active_by_var = defaultdict(list)
for var, wk in data['active_season_pairs']:
    active_by_var[var].append(wk)

print(f"=== active_season_pairs (총 {len(data['active_season_pairs'])}개) ===")
for var in data['variety_order']:
    wks = sorted(active_by_var[var])
    print(f'  - {var} ({len(wks)}주): {wks}')

holidays_sorted = sorted(pd.to_datetime(list(data['korean_holidays'])))
print(f'\n=== korean_holidays (총 {len(holidays_sorted)}일) ===')
hol_by_yr = defaultdict(list)
for dt in holidays_sorted:
    hol_by_yr[dt.year].append(dt.strftime('%Y-%m-%d'))

for yr in sorted(hol_by_yr):
    print(f'  - {yr}년 ({len(hol_by_yr[yr])}일): {hol_by_yr[yr]}')
```

## 8. 특정 품종·기간 상세 예측값 조회 예시

아래 조건(`TARGET_VARIETY`, `TARGET_FUTURE_ONLY` 등)을 변경하여 원하는 구간의 모델별 예측값을 확인할 수 있습니다.

```python
TARGET_VARIETY = '조생귤'
TARGET_FUTURE_ONLY = True

sub = all_preds_df[all_preds_df['variety'] == TARGET_VARIETY].copy()
if TARGET_FUTURE_ONLY:
    sub = sub[sub['is_future'] == True]

view_cols = [
    'date', 'group_id', 'box_unit', 'actual_vol_ton', 'actual_prc_kg',
    'RandomForest_vol_p50', 'LightGBM_vol_p50', 'XGBoost_vol_p50', 'TFT_vol_p50',
    'RandomForest_prc_p50', 'LightGBM_prc_p50', 'XGBoost_prc_p50', 'TFT_prc_p50',
]
print(f'조회 조건: variety={TARGET_VARIETY}, is_future={TARGET_FUTURE_ONLY} -> {len(sub):,}행')
display(sub[view_cols].head(20))
```
